```python
def toArrays(data):
    data_lines = data.splitlines()    
    arrays = [[int(col) for col in line.split(" ")] for line in data_lines]
    return arrays

arrays = toArrays(get_data(day=2, year=2024))
arrays[:5]
```

```python
test_data = """7 6 4 2 1
1 2 7 8 9
9 7 6 2 1
1 3 2 4 5
8 6 4 4 1
1 3 6 7 9"""

arrays = toArrays(test_data)
arrays[:5]
```

```python
def is_ascending(x):
    ascending = True
    for i in range(len(x))[1:]:
        diff = x[i] - x[i-1]
        if (x[i]>x[i-1]) & (abs(diff)<=3): 
            ascending = True
        else:
            ascending = False
            break
    return ascending
[(o, is_ascending(o)) for o in arrays]
```

```python
def is_descending(x):
    descending = True
    for i in range(len(x))[1:]:
        diff = x[i] - x[i-1]
        if (x[i]<x[i-1]) & (abs(diff)<=3): 
            descending = True
        else:
            descending = False
            break
    return descending
[(o, is_descending(o)) for o in arrays]
```

```python
[(o, is_ascending(o) | is_descending(o)) for o in arrays]
```

```python
arrays = toArrays(get_data(day=2, year=2024))
results = [(o, is_ascending(o) | is_descending(o)) for o in arrays]
results[:20]
```

```python
len(results)
```

```python
sum([o[1] for o in results])
```

```python
arrays[0]
```

```python
def is_ascending_tolerant(x):
    save_sublists = 0 
    for i in range(len(x)):
        sublist = [x[f] for f in range(len(x)) if f!=i]
        if is_ascending(sublist): 
            save_sublists+=1
    return save_sublists>0
    
[(o, (is_ascending_tolerant(o))) for o in arrays]
```

```python
def is_descending_tolerant(x):
    save_sublists = 0 
    for i in range(len(x)):
        sublist = [x[f] for f in range(len(x)) if f!=i]
        if is_descending(sublist): 
            save_sublists+=1
    return save_sublists>0
    
[(o, (is_descending_tolerant(o))) for o in arrays]
```

```python
[(o, (is_ascending_tolerant(o) | is_descending_tolerant(o))) for o in arrays]
```
