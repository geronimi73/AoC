`````python
def toArrays(x):
    lines=x.splitlines()
    return [[char for char in line] for line in lines], len(lines[0])

def reverseString(x):
    return "".join([x[i] for i in list(range(len(x)-1,-1,-1))])
    
def horizontal(x):
    return ["".join(line) for line in x]

def vertical(data, line_width):
    out = []
    for i in range(line_width): out.append("".join([line[i] for line in data]))
    return out

def singleDiagonal(data, row, col):
    out = []
    while row<len(data) and (col<line_width):
        out.append(data[row][col])
        row+=1
        col+=1
    return "".join(out)

def diagonals(data, line_width):
    topDiagonals = [singleDiagonal(data, 0, col) for col in range(line_width)]
    bottomDiagnoals = [singleDiagonal(data, row, 0) for row in range(1, line_width)]
    return topDiagonals + bottomDiagnoals

data_arrays, line_width = toArrays(data_test)
# data_arrays, line_width

data_strings = horizontal(data_arrays) \
    + vertical(data_arrays, line_width) \
    + diagonals(data_arrays, line_width)
data_strings [:5]
`````

`````python
data_arrays, line_width = toArrays(data_test)
`````

`````python
def getArms(x, row, col, width):
    arm1 = singleDiagonalTLBR(x, row, col)[:width]
    arm2 = singleDiagonalTRBL(x, row, col+width-1)[:width]
    return arm1, arm2

getArms(data_arrays, 0, 1, 3)
`````

`````python
search_string = "MAS"
arm_width = len(search_string)
count = 0 
for row in range(len(data_arrays)-arm_width+1):
    for col in range(line_width-arm_width+1):
        # print(col)
        arm1, arm2 = getArms(data_arrays, row, col, arm_width)
        hit = ((arm1 == search_string) | (reverseString(arm1) == search_string)) & \
            ((arm2 == search_string) | (reverseString(arm2) == search_string))        
        # print(arm1, arm2, hit)
        if hit: count+=1
count
`````
