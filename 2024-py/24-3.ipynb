```python
x = "xmul(2,4)%&mul[3,7]!@^do_not_mul(5,5)+mul(32,64]then(mul(11,8)mul(8,5))"
x
```

```python
import re
```

```python
def filter_valid_instructions(x):
    matches = re.findall(r'mul\(\d{1,3},\d{1,3}\)', x)
    return matches
instructions = filter_valid_instructions(x)
instructions
```

```python
x=instructions[0]
x
```

```python
def get_op(x):
    return re.findall(r'^(.*)\(', x)
get_op(x)
```

```python
x
```

```python
def get_arguments(x):
    return re.findall(r'^.*\((\d+),(\d+)\)', x)[0]
```

```python
from functools import reduce

reduce(lambda x, y: x * y, (1,2,3))
```

```python
from functools import reduce

result_sum = 0
data_test = "xmul(2,4)%&mul[3,7]!@^do_not_mul(5,5)+mul(32,64]then(mul(11,8)mul(8,5))"
for instr in filter_valid_instructions(data_test):
    op = get_op(instr)[0]
    arguments_int = [int(x) for x in get_arguments(instr)]
    result = None
    if (op=="mul"):
        result = reduce(lambda x, y: x * y, arguments_int)
    if result is not None: result_sum += result
    print(instr, get_op(instr), get_arguments(instr),result)

result_sum
```

# Part 2 

```python
data_test = "xmul(2,4)&mul[3,7]!^don't()_mul(5,5)+mul(32,64](mul(11,8)undo()?mul(8,5))"
```

```python
def filter_valid_instructions(x):
    matches = re.findall(r'mul\(\d{1,3},\d{1,3}\)|do\(\)|don\'t\(\)', x) 
    return matches
instructions = filter_valid_instructions(data_test)
instructions
```

```python
from functools import reduce

result_sum = 0
ops_enabled = True

data_test = "xmul(2,4)&mul[3,7]!^don't()_mul(5,5)+mul(32,64](mul(11,8)undo()?mul(8,5))"

for instr in filter_valid_instructions(data_test):
    op = get_op(instr)[0]
    arguments_int = None
    result = None
    if (op=="mul"):
        arguments_int = [int(x) for x in get_arguments(instr)]
        result = reduce(lambda x, y: x * y, arguments_int)
    elif (op=="do"):
        ops_enabled = True
    elif (op=="don\'t"):
        ops_enabled = False
    if result is not None and ops_enabled: result_sum += result
    print(instr, op, ops_enabled, arguments_int,result)

result_sum
```
