```python
def extract_lists(lines):
    list1=[]
    list2=[]
    for line in lines:
        col1, col2=[int(col.strip()) for col in line.split(" ", maxsplit=1)]
        list1.append(col1)
        list2.append(col2)
    return list1, list2
```

```python
list1, list2 = extract_lists(data_lines)
list1.sort()
list2.sort()
(list1[:5], list2[:5])
```

```python
diff=0
for i1, i2 in zip(list1, list2):
    diff+=abs(i1-i2)
diff
```
