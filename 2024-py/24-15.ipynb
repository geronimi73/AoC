````python
def get(m, pos): 
    x, y = pos
    return m[y][x] if -1<x<width and -1<y<height else None
get(mapp, (2,2)), get(mapp, (10,10))
````

````python
def sett(m, pos, val): 
    x, y = pos
    m[y][x] = val
````

````python
# def get_robot
def get_robot(m): return [(x,y) for x in range(width) for y in range(height) if get(m, (x, y)) == "@"][0]
get_robot(mapp)
````

````python
def display(m): print('\n'.join(''.join(row) for row in m))
display(mapp)
````

````python
movements = {
    ">": (1, 0),
    "<": (-1, 0),
    "^": (0, -1),
    "v": (0, 1)
}
def next_pos(m, mov, pos):    
    offset = movements[mov]
    return pos[0]+offset[0],pos[1]+offset[1]
next_pos(mapp, ">", (3,3))
````

````python
def is_box_mobile(m, mov, pos_box):
    assert get(m, pos_box) == "O"
    boxes = [pos_box]
    pos_current = pos_box
    while not get(m, next_pos(m, mov, pos_current)) is None:
        pos_current = next_pos(m, mov, pos_current)
        obj_current = get(m, pos_current)
        if obj_current == ".": return True, boxes
        elif obj_current == "#": return False, []
        elif obj_current == "O": 
            boxes.append(pos_current)
    return False
is_box_mobile(mapp, "v", (4,2))
````

````python
def is_box_mobile(m, mov, pos_box, boxes = None):
    if boxes is None: boxes = set()
    # print("call", pos_box)
    assert get(m, pos_box) == "[" #or get(m, pos_box) == "]"
    boxes.add(pos_box)
    
    pos_nextL, pos_nextR = next_pos(m, mov, pos_box), next_pos(m, mov, (pos_box[0]+1,pos_box[1]))
    obj_nextL, obj_nextR = get(m, pos_nextL), get(m, pos_nextR)

    # print("check", (pos_nextL, pos_nextR), (obj_nextL, obj_nextR))
    
    if obj_nextL == "." and obj_nextR==".": 
        return True, boxes
    elif obj_nextL == "#" or obj_nextR=="#": 
        return False, boxes
    elif "[" in [obj_nextL, obj_nextR] or "]" in [obj_nextL, obj_nextR]: 
        for pos, obj in zip([pos_nextL, pos_nextR],[obj_nextL, obj_nextR]):
            if obj == "]": pos = pos[0]-1,pos[1]
            elif obj == "[": pass
            else: continue
            if not pos in boxes:
                tmp_mobile, tmp_boxes  = is_box_mobile(m, mov, pos, boxes)
                if tmp_mobile: boxes.update(tmp_boxes)
                else: return False, boxes
    return True, boxes

[(x,y) for x in range(width) for y in range(height) if get(mapp, (x, y)) == "["]
data = data_test2.replace("#", "##").replace(".", "..").replace("O", "[]").replace("@", "@.")
map_raw, moves = data.split("\n\n")
moves = moves.replace("\n","")
mapp = [[col for col in row] for row in map_raw.splitlines()]

display(mapp)

print(is_box_mobile(mapp, ">", (14,7)))  # [][]## + >, should fail
print(is_box_mobile(mapp, "<", (16,7)))  # [][]## + <, should work
print(is_box_mobile(mapp, "^", (16,7)))  # [][]## + ^, should work
print(is_box_mobile(mapp, ">", (4,3)))  # [][]. + >, should work
# print(is_box_mobile(mapp, ">", (5,4)))  # [] with [][] directly above + >, should work
# print(is_box_mobile(mapp, "^", (5,4)))  # [] with [][] directly above + ^, should work
print(is_box_mobile(mapp, "v", (8,6)))  # should work

data_testcase = """####################
##....[]....[]..[]##
##............[]..##
##..[][]....[]..[]##
##....[]......[]..##
##[]##..@.[]......##
##[]....[]....[]..##
##..[][]..[]..[][]##
##........[]......##
####################"""
mapp = [[col for col in row] for row in data_testcase.splitlines()]
display(mapp)

print(is_box_mobile(mapp, "v", (8,6)))  # should fail
````
