```python
def blink(stones):
    stones_blinked = {}
    def dict_add(n, num_copies):
        if not n in stones_blinked: stones_blinked[n] = num_copies
        else: stones_blinked[n]+=num_copies
    for stone in stones:
        copies = stones[stone]
        if stone == 0:
            dict_add(1, copies)
        elif len(str(stone)) % 2 ==0:    
            stone_str = str(stone)
            dict_add(int(stone_str[:len(stone_str)//2]), copies)
            dict_add(int(stone_str[len(stone_str)//2:]), copies)
        else:
            dict_add(stone*2024, copies)
    return stones_blinked   
```

```python
data = "30 71441 3784 580926 2 8122942 0 291"
stones = {int(x):1 for x in data.split(" ")}

for i in range(75):
    stones = blink(stones)
    print(i+1, sum([stones[k] for k in stones]))
```
